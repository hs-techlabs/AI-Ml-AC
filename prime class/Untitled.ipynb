```python
import json
```

```python
def load_data(filename):
    with open (filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("show_data.json")
print(data,type(data))
```

```python
#clean & structure data
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_user=set()
    
    for user in data:
        # print(user)
        #clean rating
        raw_rating=str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=text_to_num[raw_rating]
        user["rating"]=(raw_rating)

        #Handle missing
        raw_age=user.get("age")
        if (raw_age==None):
            user["age"]=None
            
        #deduplication
        if user["name"].strip() in unique_user:
            continue

        unique_user.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
data=clean_data(data)
```

```python
#average rating
def get_insight(data):

    tot_rating=0
    for user in data:
        tot_rating+=float(user["rating"])
    print(f"average rating {tot_rating/len(data)}")


    #persentage of users poor value
    poor_rating=0
    for user in data:
        if (float(user["rating"])<3):
            poor_rating+=1

    print(f"% of users rate < 3 is {poor_rating*100/len(data)}%") 
    
```

```python
get_insight(data)
```

```python
#build recomdation
def get_recomandation(data):
    recommendation=[]
    for user in data:
        curr_rec={}
        curr_rec["name"]=user["name"]

        if (float(user["rating"])>=4):
            curr_rec["brand"]="apple"
        else:
            curr_rec["brand"]="samsung"

        recommendation.append(curr_rec)
    return recommendation
```

```python
get_recomandation(data)
```
